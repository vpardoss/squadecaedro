## Investigación sobre el transporte público en Santiago

#### Contexto y motivación:

La urbanización, se define según la RAE como la “Acción y efecto de urbanizar”, a su vez, el verbo urbanizar tiene como definición “Acondicionar una fracción de terreno y prepararlo para su uso urbano, abriendo calles y dotándolas de luz, pavimentos y demás servicios”. En la actualidad, este es un fenómeno que ocurre a nivel mundial, el cual ha ido en aumento con el paso de las décadas y se espera que se continúe acrecentando en los próximos años; según estimaciones de la ONU, a día de hoy más de un 55% de la población mundial vive en zonas urbanas, y este porcentaje será del 68% de cara al 2050. A nivel país, dicho porcentaje de población que reside en zonas urbanas corresponde a un  87.8% según datos del Censo 2017.

Según el Censo del año 2024, la población nacional está compuesta de 18.480.432 personas, de las cuales 7.400.741 viven en la Región Metropolitana; a partir de esta cifra, se comprende que hay una gran cantidad de personas que se ven afectadas en su día a día por problemas propios de la urbanización y centralización (tales como pobreza, contaminación del medioambiente, entre otros), los cuales son de gran importancia tratar para lograr mejorar la calidad de vida de las personas en la capital.

El servicio de transporte público, tiene gran significancia para el desarrollo de las actividades del día a día del ciudadano promedio, por lo que es esperable que su funcionamiento sea óptimo y seguro dada su importancia. No obstante, este servicio presenta muchas fallas que repercuten en la vida de muchísimas personas que utilizan dicho medio de transporte. El problema a tratar durante el desarrollo del presente proyecto, corresponde al de la movilidad usando como medio el transporte público, específicamente, el retraso con el que pasan los buses y la inhabilitación de estaciones de metro. En lo que respecta al Transantiago, ocupa un 50% de los viajes en la partición modal, y en promedio, los viajes tienen una duración de 46 minutos (los que en auto particular durarían 27 minutos), por lo que es importante prestar atención al problema que generan los retrasos en el paso de micros, sobretodo considerando que aumenta la inseguridad de aquel que espera en el paradero para tomar su recorrido.

En este sentido, el sistema de transporte público constituye una de las principales formas de movilización a la que recurren los santiaguinos para llevar a cabo sus responsabilidades del día a día, teniendo como opciones principales: metro, transantiago y taxi. Debido a la importancia que tiene este servicio para el individuo promedio, es importante prestar atención a sus fallas (retrasos, cierre de estaciones, etc), y buscar soluciones mediante las que se consiga un mejoramiento y optimización de este servicio de uso masivo. 

Según una encuesta realizada por la Facultad de Ingeniería y Ciencias Aplicadas de la Universidad de los Andes en el año 2023, en la que participaron 1.221 personas a lo largo del país, un 37% concuerda en que su principal problema es la poca frecuencia con que pasa el transporte, seguido de un 35% que opina que es la inseguridad, además, un 39% de los encuestados ocupan el servicio de transporte público a diario, siendo los principales usuarios las personas entre 18 y 34 años. Con dicha información, un eje al cual se puede acceder mediante la utilización de ciencia de datos (y el cual será el enfoque principal del presente proyecto) es la frecuencia con la que pasa el transporte, dado que a causa de él, la vida de los usuarios de este servicio se ve afectada en otros ámbitos, entre los que se encuentran: puntualidad para llegar a destino, sensación de seguridad (por ejemplo, en los paraderos de sectores peligrosos), planificación de actividades del día y tiempos a ocupar en estas, etc. 

Teniendo lo anterior en cuenta, la motivación del proyecto a desarrollar se basa en la detección de patrones de ineficiencia del sistema de transporte público, específicamente, del transantiago y el metro, para así aportar en la búsqueda de soluciones y todo lo que ello implica. Por lo tanto, la realización de la presente investigación le será de utilidad a todo usuario de los servicios de transporte ya especificados, especialmente a aquellos que se ven afectados por sus fallas, para que a partir de él puedan informarse sobre su medio de movilización frecuente, y de esta forma, planificar sus viajes de forma efectiva.

#### 1.- ¿Cuáles son los horarios que generan mayor desbalance con respecto a la frecuencia del transporte?

```python
# Separamos los datos por sector: 
recorrido_poniente = micro_limpia[micro_limpia['comuna'].isin(['QUINTA NORMAL', 'CERRO NAVIA', 'MAIPÚ', 'CERRILLOS',
                                                     'ESTACIÓN CENTRAL', 'PUDAHUEL', 'RENCA', 'LO PRADO',
                                                     'PADRE HURTADO'])]
recorrido_oriente = micro_limpia[micro_limpia['comuna'].isin(['PROVIDENCIA', 'ÑUÑOA', 'LA REINA', 'LAS CONDES',
                                                                  'VITACURA', 'LO BARNECHEA', 'PEÑALOLÉN', 'MACUL',
                                                                  'LA FLORIDA', 'PUENTE ALTO'])]
recorrido_sur = micro_limpia[micro_limpia['comuna'].isin(['PEDRO AGUIRRE CERDA', 'SAN MIGUEL',
                                                              'SAN JOAQUÍN', 'LO ESPEJO', 'LA CISTERNA',
                                                              'LA GRANJA', 'SAN RAMÓN', 'EL BOSQUE',
                                                              'SAN BERNARDO', 'LA CISTERNA'])]
recorrido_norte = micro_limpia[micro_limpia['comuna'].isin(['CONCHALÍ', 'HUECHURABA', 'INDEPENDENCIA',
                                                                         'RECOLETA', 'QUILICURA'])]

print("CANTIDAD DE DATOS DE LAS COMUNAS DEL SECTOR ORIENTE:")
valores_poniente = recorrido_poniente["comuna"].value_counts()
valores_oriente = recorrido_oriente["comuna"].value_counts()
valores_norte = recorrido_norte["comuna"].value_counts()
valores_sur = recorrido_sur["comuna"].value_counts()
for indice,value in enumerate(valores_poniente):
    print(f"{valores_poniente.index[indice]}: {value}", end=" ")
print("\nCANTIDAD DE DATOS EN LAS COMUNAS DEL SECTOR ORIENTE")
for indice,value in enumerate(valores_oriente):
    print(f"{valores_oriente.index[indice]}: {value}", end=" ")
print("\nCANTIDAD DE DATOS EN LAS COMUNAS DEL SECTOR NORTE")
for indice,value in enumerate(valores_norte):
    print(f"{valores_norte.index[indice]}: {value}", end=" ")
print("\nCANTIDAD DE DATOS EN LAS COMUNAS DEL SECTOR SUR")
for indice,value in enumerate(valores_sur):
    print(f"{valores_sur.index[indice]}: {value}", end=" ")
```

```python
# Revisaremos la mediana del tiempo mínimo de espera por sector
mediana_time_norte = recorrido_norte.groupby(by='hour').agg({'min_arrival_time': 'median'})
mediana_time_sur = recorrido_sur.groupby(by='hour').agg({'min_arrival_time': 'median'})
mediana_time_oriente = recorrido_oriente.groupby(by='hour').agg({'min_arrival_time': 'median'})
mediana_time_poniente = recorrido_poniente.groupby(by='hour').agg({'min_arrival_time': 'median'})
mediana_oriente_poniente = pd.merge(mediana_time_oriente, mediana_time_poniente, on='hour', how='outer')
mediana_oriente_poniente = mediana_oriente_poniente.set_axis(['Sector Oriente', 'Sector Poniente'], axis= 1)
mediana_norte_sur = pd.merge(mediana_time_norte, mediana_time_sur, on='hour', how='outer')
mediana_norte_sur = mediana_norte_sur.set_axis(['Sector Norte', 'Sector Sur'], axis= 1)
mediana_time_sector = pd.merge(mediana_norte_sur, mediana_oriente_poniente, on='hour', how='outer')
mediana_time_sector = mediana_time_sector.sort_index()

sectores = mediana_time_sector.columns
n_sectores = len(sectores)

fig, axes = plt.subplots(2, 2, figsize=(14, 10), sharex=True, sharey=True)
axes = axes.flatten()
colormap = plt.colormaps['tab10']

for i, columna in enumerate(sectores):
    ax = axes[i]
    ax.plot(mediana_time_sector.index,
            mediana_time_sector[columna],
            marker='o',
            linewidth=2,
            color=colormap(i))

    ax.set_title(f"Sector: {columna}")
    ax.set_xlabel("Hora")
    ax.set_ylabel("Mediana de tiempo de espera")
    ax.grid(True, linestyle='-', alpha=0.6)

fig.suptitle("Mediana de tiempo mínimo de espera por hora y sector", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

Los horarios más problemáticos corresponden a las 6 am y a las 22 hrs, siendo esta última hora poco mayor a la mediana de las otras horas en todos los sectores, ahora bien, es importante revisar qué está pasando en cada sector específicamente:
* Sector norte: Mediana mayoritariamente comprendida entre los 8-9 minutos, aquí a las 22hrs no vemos una irregularidad mayor que a la mayoría de horas del gráfico, pero al observar lo que pasa a las 6am, podemos ver un gran incremento en el tiempo que tardan en pasar las micros.
* Sector sur: Es interesante lo que ocurre en este sector, primeramente, es el único sector en el que la mediana de tiempo de espera no aumenta significativamente con respecto a las otras horas; también podemos ver que por lo general el tiempo de espera es constante al paso de las horas, y los recorridos que pasan a las 22:00hrs sólo tardan 1 minuto más de la moda de medianas. Podríamos decir que es el sector con mejores números en este ámbito
* Sector Oriente: En este caso, la mediana de tiempo mínimo de espera es superior a las 6:00hrs (por 2 minutos) y a las 22:00hrs (por 1 minuto), en el resto de horas, el comportamiento es constante y se queda en 8 minutos.
* Sector Poniente: En el último gráfico revisado, notamos que a las 06:00hrs la mediana de tiempo de espera es bastante superior con respecto a las otras horas y también con respecto a los otros gráficos, a su vez, vemos que las 22:00hrs no es una hora probblemática porque no se sale del marco general de mediana de tiempos de espera. Finalmente, notamos que este sector tiene como constante una mediana de 9 minutos en la mayoría de horarios, entonces al observar los datos del gráfico y su comportamiento, hace que sea el sector con peores números en este sentido.

Ahora, es importante convenir en que la diferencia entre cada gráfico no es mayormente exorbitante, pero al tratarse del transporte público, el tiempo se vuelve importantísimo, sobretodo a las 06:00hrs (hora punta) y a las 22:00hrs (hora arriesgada para estar en un paradero), por lo tanto implica que hay que trabajar en ello para ofrecer mayor seguridad a los usuarios del Transantiago.

#### 3.- ¿La frecuencia del transporte por sector es proporcional a la densidad demográfica respectiva?

```python
#importamos la información de las proyecciones poblacionales del censo y la filtramos por año y region
df_densidad = pd.read_csv('../data/censo_proyecciones_año.csv', sep = ";")
df_densidad = df_densidad[df_densidad['region'] == "Metropolitana de Santiago"]
df_densidad = df_densidad[df_densidad['año'] == 2025]
df_densidad = df_densidad.sort_values(by='población')
df_densidad.head()
```

```python
#graficamos la densidad poblacional
gdf = gpd.read_file("../data/comunas/comunas.shp")
```

```python
df_densidad = df_densidad[df_densidad['region'] == "Metropolitana de Santiago"]
gdf = gdf[gdf['Region'] == "Región Metropolitana de Santiago"]
gdf = gdf.merge(df_densidad, left_on="Comuna", right_on="comuna", how="left")
```

```python
gdf = gdf.to_crs(epsg=3857) 
gdf['area_km2'] = gdf.geometry.area / 10**6
gdf['densidad'] = gdf['población'] / gdf['area_km2']
```

```python
fig, ax = plt.subplots(1, 1, figsize=(12, 12))
gdf.plot(column='densidad', cmap='OrRd', linewidth=0.8, ax=ax, edgecolor='0.8', legend=True)
ax.set_title("Densidad poblacional por comuna – Región Metropolitana")
ax.axis('off')
plt.show()
```

```python
df_micros = pd.read_csv("../data/paraderos.csv")
df_grouped = df_micros.groupby("comuna", as_index=False)["n_servicios"].sum()
```

```python
# Apartamos las comunas con su respectiva población y estandarizamos los nombres de las comunas
df_comunas = df_densidad[['comuna', 'población']].copy()
df_comunas['comuna'] = (df_comunas['comuna'].str.normalize("NFD").str.encode("ascii", "ignore").str.decode("utf-8").str.lower())
#Aplicamos la misma estandarización para la cantidad de micros por comuna
df_grouped['comuna'] = (df_grouped['comuna'].str.normalize("NFD").str.encode("ascii", "ignore").str.decode("utf-8").str.lower())

#Unimos
df_unido = df_comunas.merge(df_grouped, on="comuna", how="outer")

#Botamos filas que no corresponden a Santiago
df_unido = df_unido.dropna(subset=["n_servicios", "población"])
```

```python
#Graficamos usando la recta x = y para tener como referencia una proporcionalidad estricta
plt.figure(figsize=(8,5))

plt.scatter(df_unido["población"], df_unido["n_servicios"])

plt.xlabel("Cantidad de personas")
plt.ylabel("Número de servicios")
plt.title("Relación entre población y número total de servicios por comuna")
plt.grid(True)

x_min, x_max = plt.xlim()
y_min, y_max = plt.ylim()

m = (y_max - y_min) / (x_max - x_min)
b = y_min - m * x_min

x_vals = np.linspace(x_min, x_max, 200)
plt.plot(x_vals, m * x_vals + b, '--')

plt.show()
```

Al observar el gráfico, es posible notar que existe una correlación positiva entre la población total por comuna y el número de servicios disponibles, lo que nos permite decir que a mayor número de personas hayan en las comunas, estas contarán con mayor cantidad de servicios de micros. 

A partir de esto, podemos pensar que es un factor que se considera al momentto de determinar la cobertura de transporte público para las comunas de la Región Metropolitana.

A partir de ello, podemos esperar una mayor oferta de buses que en las horas valle/noche. Veamos si esto se cumple:

```python
disponibilidad = [0.9, 1.22, 1.01, 0.73]  # servicios por hora
calidad = [1.0, 1.0, 0.44, 1.0]            # tiempos de espera
labels = ["am/valle", "pm/valle_pm", "pm/noche", "am/pm"]

x = np.arange(len(labels))
plt.figure(figsize=(10, 5))

# graficar variable cocientes
plt.bar(x - 0.15, disponibilidad, 0.3, label="Disponibilidad", 
        color="rosybrown", alpha=0.8)
# graficar calidad_transportes
plt.bar(x + 0.15, calidad, 0.3, label="Calidad en tiempo", 
        color="chocolate", alpha=0.8)


plt.axhline(y=1.0, color="red", linestyle="--", linewidth=1, 
            label="mínima esperada", alpha=0.7)
plt.xlabel("Comparación entre períodos")
plt.ylabel("Ratio")
plt.title("Proporcionalidad del Servicio de Transporte")
plt.xticks(x, labels)
plt.legend()

for i, (disp, cal) in enumerate(zip(disponibilidad, calidad)):
    plt.text(i - 0.15, disp + 0.05, f"{disp:.2f}", 
             ha="center", va="bottom", fontsize=9)
    plt.text(i + 0.15, cal + 0.05, f"{cal:.2f}", 
             ha="center", va="bottom", fontsize=9)

plt.grid(True, alpha=0.2, axis="y")
plt.ylim(0, 1.5)
plt.tight_layout()
plt.show()
```
